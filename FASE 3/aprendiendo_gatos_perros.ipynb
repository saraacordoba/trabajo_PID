# 1. Llevar el conjunto de datos al entorno

```python
transform = transforms.Compose([
    transforms.RandomHorizontalFlip(), # voltea la imagen en efecto espejo
    transforms.RandomRotation(20), # gira la imagen 20 grados
    transforms.Resize((64, 64)),
    transforms.ToTensor(), 
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
    ])
```

```python
train_data = datasets.ImageFolder('gatos_perros', transform=transform)
```

```python
print(f"Total imágenes en el dataset: {len(train_data)}")
print(f"Mapeo de clases: {train_data.class_to_idx}")
```

```python
test_size = 0.2
valid_size = 0.2
batch_size = 16 # cuántas imágenes se procesan juntas a la vez

# For test and train
num_data = len(train_data)
indices_data = list(range(num_data)) # una lista con la posicion de cada imagen
np.random.shuffle(indices_data)
split_tt = int(np.floor(test_size * num_data)) 
train_idx, test_idx = indices_data[split_tt:], indices_data[:split_tt] 

# For Valid
num_train = len(train_idx)
np.random.shuffle(train_idx)
split_tv = int(np.floor(valid_size * num_train))
train_final_idx = train_idx[split_tv:]
valid_idx = train_idx[:split_tv]
train_idx = train_final_idx 


# define samplers for obtaining training and validation batches
train_sampler = SubsetRandomSampler(train_idx)
print(list(train_sampler))
test_sampler = SubsetRandomSampler(test_idx)
valid_sampler = SubsetRandomSampler(valid_idx)


# Loaders contains the data in tuple format 
# (Image in form of tensor, label)

# - Imágenes: bloque numérico de 4 dimensiones que agrupa las fotos del lote: {Forma (Shape):  [batch_size, canales, alto, ancho]
# - Etiquetas:  lista con la clase real que corresponde a cada una de las 16 imágenes del paquete: [0,1,1,0,...]
train_loader = torch.utils.data.DataLoader(train_data, batch_size=batch_size, sampler=train_sampler, num_workers=0)
valid_loader = torch.utils.data.DataLoader(train_data, batch_size=batch_size, sampler=valid_sampler, num_workers=0)
test_loader = torch.utils.data.DataLoader(train_data, batch_size=batch_size, sampler=test_sampler, num_workers=0)

# variable representing classes of the images
classes = [0, 1]
```

# 2. Inicialización de la CNN

La CNN en PyTorch se define de la siguiente manera:

torch.nn.Conv2d(Profundidad_de_imagen_entrada, Profundidad_de_filtro, tamaño_de_filtro, relleno, pasos)

- profundidad de la imagen de entrada es generalmente 3 para RGB y 1 para escala de grises. D
- profundidad del filtro: cuántos detectores de características (filtros) independientes aplicará la capa.
- tamaño de filtro: tamaño de la ventana/matriz que escanea la imagen.
- padding: Bordes de ceros que se añaden alrededor de la imagen para que los filtros puedan procesar bien los bordes sin reducir demasiado el tamaño. Por defecto suele ser 0 o 1.
- stride (pasos): De cuántos en cuántos píxeles se desplaza el filtro a lo largo de la imagen. Por defecto es 1.

Cada vez que aplicas una capa convolucional, el tamaño espacial (alto y ancho) de la imagen cambia según esta fórmula:$$D_{out} = \left\lfloor \frac{D_{in} - K + 2P}{S} \right\rfloor + 1$$
Donde:
- $D_{in}$: Dimensión de entrada de la imagen (en este caso, 64 píxeles).
- $K$: kernel_size (tamaño del filtro, ej. 3).
- $P$: padding (relleno, ej. 1).
- $S$: stride (paso, ej. 1).

Inicializamos el modelo:

```python
class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        # convolutional layer
        self.conv1 = nn.Conv2d(3, 16, 5) 
        # max pooling layer
        self.pool = nn.MaxPool2d(2, 2)  
        self.conv2 = nn.Conv2d(16, 32, 5)
        self.dropout = nn.Dropout(0.2)  
        # NOTA
        self.fc1 = nn.Linear(32 * 13 * 13, 256)
        self.fc2 = nn.Linear(256, 84)
        self.fc3 = nn.Linear(84, 2) 
        self.softmax = nn.LogSoftmax(dim=1)
        
    def forward(self, x): 
       
        x = self.pool(F.relu(self.conv1(x)))  
        x = self.pool(F.relu(self.conv2(x)))  
        x = self.dropout(x) 
        x = x.view(-1, 32 * 13 * 13) 
      
        x = F.relu(self.fc1(x))
        x = self.dropout(F.relu(self.fc2(x)))
        x = self.softmax(self.fc3(x))
        return x
    
# create a complete CNN
model = Net()

# Loss function: encargada de medir qué tan mal o bien está prediciendo la red. Compara la salida del modelo con la respuesta real.
# criterion = torch.nn.CrossEntropyLoss()
criterion = torch.nn.NLLLoss()
# Optimizer: Descenso de Gradiente Estocástico.
optimizer = torch.optim.SGD(model.parameters(), lr = 0.003, momentum= 0.9) # lr = tasa de aprendizaje. momentum: no entiendo bien 
```

## NOTA
¿Cómo cambia el tamaño de una imagen cuando pasa por una capa convolucional?
$$\text{Nuevo Tamaño} = \left\lfloor \frac{\text{Tamaño Entrada} - \text{Kernel} + 2 \times \text{Padding}}{\text{Stride}} \right\rfloor + 1$$


Con mi imágen:  ($64 \times 64$):
- Paso 1: Primera Convolución (conv1): aplicamos la fórmula con entrada 64 y Kernel 5:$$\text{Tamaño} = \frac{64 - 5 + 2(0)}{1} + 1 = 59 + 1 = 60$$Pasa de $64 \times 64$ a $60 \times 60$.
- Paso 2: Primer Max Pooling (pool): divide la resolución espacial entre 2:$$\text{Tamaño} = \frac{60}{2} = 30$$Pasa a ser $30 \times 30$.- Paso 3: Segunda Convolución (conv2): aplicamos de nuevo la fórmula con entrada 30 y Kernel 5:$$\text{Tamaño} = \frac{30 - 5 + 2(0)}{1} + 1 = 25 + 1 = 26$$Pasa a ser $26 \times 26$.
- Paso 4: Segundo Max Pooling (pool): volvemos a dividir entre 2:$$\text{Tamaño} = \frac{26}{2} = 13$$Pasa a ser $13 \times 13$.
- 32 es el número de filtros (o canales de salida) de la última capa convolucional

# 3. Entrenar el modelo

Los 6 pasos del entrenamiento por cada lote (Batch):
1. Borrar gradientes (optimizer.zero_grad()):
PyTorch acumula los gradientes por defecto en cada iteración. Si no los borras a cero antes de procesar un nuevo lote, sumará las correcciones del lote actual con las del anterior y el modelo no aprenderá correctamente.

2. Paso hacia adelante / Forward Pass (output = model(data)):
Las imágenes del lote entran a la red y se calcula la predicción actual.

3. Calcular la pérdida / Loss (loss = criterion(output, target)):
Compara la predicción devuelta por la red (output) con la etiqueta real (target) para ver cuánto se ha equivocado.

4. Paso hacia atrás / Backward Pass (loss.backward()):
Calcula las derivadas (gradientes) del error hacia atrás a través de toda la red para saber cuánto ha contribuido cada peso al fallo.

5. Optimización (optimizer.step()):
El optimizador aplica el algoritmo para actualizar y corregir ligeramente todos los pesos de la red.

6. Acumular el error (train_loss += loss.item() * data.size(0)):
Suma el error ponderado del lote para al final del ciclo poder saber la pérdida media de toda la época.

```python
# Definimos la variable manualmente
train_on_gpu = torch.cuda.is_available()

# Si hay GPU, también enviamos el modelo a la tarjeta gráfica
if train_on_gpu:
    print('CUDA disponible. El entrenamiento se ejecutará en la GPU.')
    model.cuda()
else:
    print('CUDA no disponible. El entrenamiento se ejecutará en la CPU.')
```

```python
# number of epochs to train the model
n_epochs = 5 

valid_loss_min = np.inf # guarda la MEJOR perdida minima de validación

for epoch in range(1, n_epochs+1):

    # acumuladores a 0
    train_loss = 0.0
    valid_loss = 0.0
    
    # 1. ENTRENAR
    model.train() # modo entrenamiento. El dropout se activa
    
    for data, target in train_loader: 

        # Si hay GPU, mueve el lote allí.
        if train_on_gpu:
            data, target = data.cuda(), target.cuda()

  
        optimizer.zero_grad() # 1
        output = model(data) # 2
        loss = criterion(output, target) # 3
        loss.backward() # 4
        optimizer.step() # 5
        # update training loss
        train_loss += loss.item()*data.size(0) # Apunta el error para imprimirlo luego
        
    # 2. VALIDAR
    model.eval() # dropout se desactiva

    for data, target in valid_loader:

        if train_on_gpu:
            data, target = data.cuda(), target.cuda()

        # lo mismo. Solo mide, no aprende
        output = model(data)
        loss = criterion(output, target)
        valid_loss += loss.item()*data.size(0)

    
    # 3. RESUMIR
    train_loss = train_loss / len(train_sampler) # divide el error que ha ido acumulando cada lote entre el numero de imagenes
    valid_loss = valid_loss / len(valid_sampler)
        
    # Imprime como ha ido la época
    print('Epoch: {} \tTraining Loss: {:.6f} \tValidation Loss: {:.6f}'.format(
        epoch, train_loss, valid_loss))
    
    # guarda el mejor modelo
    if valid_loss <= valid_loss_min:
        print('Validation loss decreased ({:.6f} --> {:.6f}).  Saving model ...'.format(
        valid_loss_min,
        valid_loss))
        torch.save(model.state_dict(), 'prueba.pt')
        valid_loss_min = valid_loss


        
```

# 4. Evaluación del modelo

```python
# 1. Variables de control
test_loss = 0.0
class_correct = [0.0, 0.0] # aciertos de [clase 0, clase 1]
class_total = [0.0, 0.0]   # imágenes vistas de [clase 0, clase 1]
classes = ['fake', 'real']
batch_size = 16 

model.eval() # Modo evaluación

# Desactivamos el cálculo de gradientes para ahorrar memoria y acelerar la evaluación
with torch.no_grad():

    for data, target in test_loader:
        if train_on_gpu:
            data, target = data.cuda(), target.cuda()
            
        output = model(data)
        loss = criterion(output, target)
        test_loss += loss.item() * data.size(0)

        # --------------------------------------------------------------- APUNTES
        # Obtener la clase predicha
        _, pred = torch.max(output, 1)
        correct_tensor = pred.eq(target.data.view_as(pred))
        
        if train_on_gpu:
            correct = np.squeeze(correct_tensor.cpu().numpy())
        else:
            correct = np.squeeze(correct_tensor.numpy())
            
        # Acumular aciertos por clase
        for i in range(len(target)):
            label = target.data[i]
            class_correct[label] += correct[i].item() if correct.ndim > 0 else correct.item() # para un caso raro: si un lote tiene una sola imagen, squeeze convierte [True] en True a secas, y entonces correct[i] fallaría.
            class_total[label] += 1




# 2. Calcular pérdidas y precisión globales
# test_loss = test_loss / len(test_loader.dataset)
test_loss = test_loss / len(test_sampler)
print(f'Test Loss: {test_loss:.6f}\n')

# Precisiones por cada clase individual 0 Y 1
for i in range(2):
    if class_total[i] > 0:
        acc = 100 * class_correct[i] / class_total[i]
        print(f'Test Accuracy of {classes[i]:>5s}: {acc:2.0f}% ({int(class_correct[i])}/{int(class_total[i])})')
    else:
        print(f'Test Accuracy of {classes[i]:>5s}: N/A (sin ejemplos)')

# Precisión global total ACCURACY
total_correct = np.sum(class_correct)
total_samples = np.sum(class_total)
overall_acc = 100. * total_correct / total_samples
print(f'\nTest Accuracy (Overall): {overall_acc:2.0f}% ({int(total_correct)}/{int(total_samples)})')
```
